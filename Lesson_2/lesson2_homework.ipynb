```python
# Блок 1: Список з 8 записів про видачу книжок в бібліотеці
requests = [
    {"id": 1, "service": "художня", "wait_days": 5, "resolved": True},
    {"id": 2, "service": "підручники", "wait_days": 12, "resolved": False},
    {"id": 3, "service": "наукова", "wait_days": 2, "resolved": True},
    {"id": 4, "service": "художня", "wait_days": 9, "resolved": True},
    {"id": 5, "service": "довідники", "wait_days": 1, "resolved": True},
    {"id": 6, "service": "підручники", "wait_days": 15, "resolved": False},
    {"id": 7, "service": "наукова", "wait_days": 3, "resolved": True},
    {"id": 8, "service": "художня", "wait_days": 11, "resolved": False},
]

print("Загальна кількість записів:", len(requests))
print("Категорія першого запису:", requests[0]["service"])
```

```python
# Блок 2: Розрахунок трьох показників за допомогою циклу for
total = len(requests)
resolved_count = 0
wait_sum = 0

for r in requests:
    wait_sum = wait_sum + r["wait_days"]
    if r["resolved"]:
        resolved_count = resolved_count + 1

print("Всього видач книжок:", total)
print("Частка повернутих книжок, %:", round(resolved_count / total * 100, 1))
print("Середня затримка, днів:", round(wait_sum / total, 1))
```

```python
# Блок 3: Класифікація затримки повернення
for r in requests:
    if r["wait_days"] <= 3:
        label = "швидко"
    elif r["wait_days"] <= 10:
        label = "нормально"
    else:
        label = "довго"
    print(r["id"], r["service"], label)
```

```python
# Блок 4: Функції з return, додаткова функція та перевірки assert

def average_wait(records):
    total = 0
    for r in records:
        total = total + r["wait_days"]
    return total / len(records)

def classify_wait(days):
    if days <= 3:
        return "швидко"
    elif days <= 10:
        return "нормально"
    else:
        return "довго"

# Додаткова (4-та) функція: пошук найчастішої категорії
def most_frequent_service(records):
    counts = {}
    for r in records:
        s = r["service"]
        counts[s] = counts.get(s, 0) + 1
    return max(counts, key=counts.get)

# Перевірки assert
assert classify_wait(3) == "швидко"
assert classify_wait(10) == "нормально"
assert round(average_wait(requests), 1) == 7.2
assert most_frequent_service(requests) == "художня"

print("Найчастіша категорія книжок:", most_frequent_service(requests))
print("Усі перевірки (включно з додатковою функцією) пройдено успішно!")
```
