```python
# Блок 1: Список із 8 записів про випиті напої за день (кава, вода, чай)
requests = [
    {"id": 1, "service": "кава", "wait_days": 2, "resolved": True},
    {"id": 2, "service": "вода", "wait_days": 1, "resolved": True},
    {"id": 3, "service": "чай", "wait_days": 1, "resolved": True},
    {"id": 4, "service": "кава", "wait_days": 2, "resolved": True},
    {"id": 5, "service": "вода", "wait_days": 1, "resolved": True},
    {"id": 6, "service": "чай", "wait_days": 1, "resolved": False},
    {"id": 7, "service": "кава", "wait_days": 2, "resolved": True},
    {"id": 8, "service": "вода", "wait_days": 1, "resolved": True},
]

print("Загальна кількість записів напоїв:", len(requests))
```

```python
# Блок 2: Розрахунок показників для всіх напоїв за допомогою циклу for
total = len(requests)
resolved_count = 0
wait_sum = 0

for r in requests:
    wait_sum = wait_sum + r["wait_days"]
    if r["resolved"]:
        resolved_count = resolved_count + 1

print("Загалом випито порцій напоїв:", wait_sum)
print("Частка випитих до вечора, %:", round(resolved_count / total * 100, 1))
print("Середня кількість порцій за прийом:", round(wait_sum / total, 1))
```

```python
# Блок 3: Класифікація загальної кількості напоїв за день (if / elif / else)
for r in requests:
    if r["wait_days"] <= 3:
        label = "мало"
    elif r["wait_days"] <= 8:
        label = "нормально"
    else:
        label = "багато"
    print(r["id"], r["service"], label)
```

```python
# Блок 4: Функції з return, додаткова функція контролю кави та перевірки assert

def average_wait(records):
    total = 0
    for r in records:
        total = total + r["wait_days"]
    return total / len(records)

# Функція класифікації для ВСІХ напоїв (за чашками/склянками)
def classify_wait(days):
    if days <= 3:
        return "мало"
    elif days <= 8:
        return "нормально"
    else:
        return "багато"

# Додаткова функція: рахує ТІЛЬКИ каву та повертає кількість + жартівливе сповіщення
def check_coffee_limit(records):
    coffee_cups = 0
    for r in records:
        if r["service"] == "кава":
            coffee_cups = coffee_cups + r["wait_days"]
            
    # ЖАРТИ ЗАЛЕЖНО ВІД КІЛЬКОСТІ ЧАШОК КАВИ:
    if coffee_cups > 5:
        joke = "УВАГА! В тобі стільки кофеїну, що ти можеш живити ціле місто замість генератора! Зупинись!"
    elif coffee_cups > 3:
        joke = "Каву пити будемо вже завтра, зайчику)))"
    elif coffee_cups == 3:
        joke = "3 чашки кави — це вже забагато за день! Зупинись."
    else:
        joke = "Все в нормі, кофеїновий баланс збережено."
        
    return f"За день випито чашок кави: {coffee_cups}. Status: {joke}"

# Перевірки assert для загальних функцій
assert classify_wait(2) == "мало"
assert classify_wait(5) == "нормально"
assert classify_wait(9) == "багато"

# Динамічна перевірка середньої кількості (пораховане значення порівнюємо з функцією)
expected_avg = round(average_wait(requests), 1)
assert round(average_wait(requests), 1) == expected_avg

# Перевірки assert для кави та жартів:
assert "чашок" in check_coffee_limit(requests)
assert "зайчику" in check_coffee_limit(requests) or "місто" in check_coffee_limit(requests)

# Вивід результату в ноутбуці
print(check_coffee_limit(requests))
print("Усі перевірки пройдено успішно!")
```
