```python
# Залишаємо тільки унікальні партії (35 000 замість 35 700)
df_clean = df.drop_duplicates(subset=["lot_id"])
print("Розмір без повторів партій:", df_clean.shape)
```

```python
# Перетворюємо рік на тип дати
df_clean["harvest_calendar_year"] = pd.to_datetime(df_clean["harvest_calendar_year"], format="%Y")

# Перевіряємо тип стовпчика і найраніший/найпізніший рік
print("Тип даних стовпчика дати:", df_clean["harvest_calendar_year"].dtype)
print("Мінімальний рік:", df_clean["harvest_calendar_year"].dt.year.min())
print("Максимальний рік:", df_clean["harvest_calendar_year"].dt.year.max())
```

```python
# Шукаємо аномально високі ферми (більше 2500 метрів)
print("Ферми з висотою > 2500м:", len(df_clean[df_clean["altitude_mean_meters"] > 2500]))

# Шукаємо дуже старі дерева (більше 40 років)
print("Дерева віком > 40 років:", len(df_clean[df_clean["tree_age_years"] > 40]))
```

```python
# 1. Знаходимо медіану щільності зерен
med_density = df_clean["bean_density_g_ml"].median()

# 2. Заповнюємо пропуски медіаною
df_clean["bean_density_filled"] = df_clean["bean_density_g_ml"].fillna(med_density)

# 3. Перевіряємо, чи залишилися пропуски в новому стовпчику
print("Пропусків до заповнення:", df_clean["bean_density_g_ml"].isna().sum())
print("Пропусків після заповнення:", df_clean["bean_density_filled"].isna().sum())
```

```python
# 1. Побудова підсумкового звіту про пропуски, унікальні значення та типи
report = pd.DataFrame({
    "пропуски": df_clean.isna().sum(),
    "унікальні": df_clean.nunique(),
    "тип_даних": df_clean.dtypes.astype(str)
})
print("=== Звіт про якість даних (перші 10 стовпчиків) ===")
print(report.head(10))

# 2. Приватність та захист даних: видаляємо технічні ідентифікатори
safe_df = df_clean.drop(columns=["record_id", "lot_id"])
print("\nКількість стовпчиків у безпечній копії safe_df:", len(safe_df.columns))
```

```python
import pandas as pd

# 1. Загружаем исходный файл (создаём df)
df = pd.read_csv("coffee_value_benchmark.csv")

# 2. Удаляем дубликаты партий (создаём df_clean)
df_clean = df.drop_duplicates(subset=["lot_id"])

# 3. Код преподавателя: удаляем технические ID и сохраняем в CSV
safe_df = df_clean.drop(columns=["record_id", "lot_id"])
safe_df.to_csv("coffee_clean.csv", index=False)

print("Успішно збережено!")
```
